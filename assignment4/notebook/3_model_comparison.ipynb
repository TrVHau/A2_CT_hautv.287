## 1. Load kết quả từ 2 Notebook trước

```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import os
from pathlib import Path

BASE_DIR = Path.cwd().parent
# Sửa lại đường dẫn cho đúng vị trí thực tế của từng file
dogcat_file = BASE_DIR / "dogcat_results.csv"
mnist_file = BASE_DIR / "notebook/mnist_results.csv"

has_dogcat = dogcat_file.exists()
has_mnist = mnist_file.exists()

print("Có kết quả Dog/Cat:", has_dogcat)
print("Có kết quả MNIST:", has_mnist)

df_dogcat = pd.read_csv(dogcat_file) if has_dogcat else None
df_mnist = pd.read_csv(mnist_file) if has_mnist else None
```

## 2. So sánh kết quả Dog/Cat

```python
if has_dogcat:
    display(df_dogcat)
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].bar(df_dogcat['model'], df_dogcat['test_accuracy'], color=['#4C78A8', '#F58518', '#54A24B'])
    axes[0].set_title('Test Accuracy - Dog/Cat')
    axes[0].set_ylim(0, 1)

    axes[1].bar(df_dogcat['model'], df_dogcat['test_loss'], color=['#4C78A8', '#F58518', '#54A24B'])
    axes[1].set_title('Test Loss - Dog/Cat')
    plt.show()
```

## 3. So sánh kết quả MNIST

```python
if has_mnist:
    display(df_mnist)
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].bar(df_mnist['model'], df_mnist['test_accuracy'], color=['#4C78A8', '#F58518', '#54A24B'])
    axes[0].set_title('Test Accuracy - MNIST')
    axes[0].set_ylim(0, 1)

    axes[1].bar(df_mnist['model'], df_mnist['test_loss'], color=['#4C78A8', '#F58518', '#54A24B'])
    axes[1].set_title('Test Loss - MNIST')
    plt.show()
```

### Sự khác biệt giữa Dog/Cat và MNIST
1. **MNIST dễ hơn rất nhiều:** 
   - Ảnh là grayscale (1 kênh màu), ảnh nhỏ (28x28).

   - Đối tượng có hậu cảnh phức tạp, góc chụp khác nhau, con vật đổi tư thế, che khuất...
   - Mô hình CNN cơ bản chỉ đạt **65-75%**, cần dùng mạng pretrained (VGG, ResNet) hoặc tăng cường dữ liệu cường độ lớn để đạt >90%.
